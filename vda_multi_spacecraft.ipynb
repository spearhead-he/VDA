# SPEARHEAD VDA tool: multiple spacecraft

The tool performs the Velocity Dispersion Analysis (VDA) of one Solar Energetic Particle event, given as a datetime range, from the data of several spacecraft, and outputs the release time and apparent path length of the event from each spacecraft, with their VDA fits in one plot.

The spacecraft are chosen in the Spacecraft tab, from **Solar Orbiter**, **STEREO-A**, **Parker Solar Probe**, **SOHO**, **Wind** and **BepiColombo** (see [Observers](https://github.com/spearhead-he/VDA/blob/main/docs/user_guide.md#observers) for their instruments and viewings). Each spacecraft has its own analysis, as in the multi-event notebook (`vda_multi_event.ipynb`); the event, the onset determination and selection, and the plots are the same for all of them.

Run the cells in order: set the parameters in the form, then run the steps of the analysis. If a parameter is changed, rerun the steps after it. See the [user guide](https://github.com/spearhead-he/VDA/blob/main/docs/user_guide.md#multiple-spacecraft) for details.

```python
try:
    from spearhead.vda.notebook import VDA_multi_spacecraft_notebook
except ModuleNotFoundError:
    # running from the repository without installing the package
    import sys
    sys.path.insert(0, "src")
    from spearhead.vda.notebook import VDA_multi_spacecraft_notebook

tool = VDA_multi_spacecraft_notebook()
```

## Parameters

- **Event**: the datetime range of the data of the event and the folder of the downloaded data files (by default `particle_data` in the folder of the notebook), the same for all the spacecraft.
- **Spacecraft**: the spacecraft whose data are used, and in a tab per spacecraft its viewings, resample frequency ([offset alias](https://pandas.pydata.org/pandas-docs/stable/user_guide/timeseries.html#offset-aliases), e.g. `5min`, blank for no resampling) and grouped energy channels ("Add Channel" / "Remove Channel"), starting from the defaults of the spacecraft. *The order of the viewings is the priority of the "Use all" onset selection.*
- **Onsets**: the onset determination method (Sigma threshold or Poisson-CUSUM) and its parameters, the default background window (in minutes after the start of the event), and the onset selection method:
  - Use all: for each grouped channel, the onset of the first viewing (in the order of the viewings) with a determined onset is used
  - Interactive: the onsets are shown one channel at a time, for the spacecraft chosen in a dropdown, and the viewing whose onset is used is selected per channel (or none)
- **Views/Plots**:
  - font sizes, and saving of the VDA plots as .png files
  - whether the VDA plot of each spacecraft is also shown
  - reference times (e.g. SXR peak, type II and type III radio bursts): a label, the observed datetime and the observer (Earth or a spacecraft, or its distance from the Sun). They are shown at inverse beta 0 at the Sun, i.e. minus the light travel time from the Sun to the observer
  - display of the intermediate tables (off by default)

```python
tool.parameters_form()
```

## Data

Downloads the data of each spacecraft and groups their energy channels. A spacecraft without data for the event is left out of the next steps, with its error.

```python
tool.run_data()
```

## Background windows

Shows the event with its background window, for the spacecraft chosen in the dropdown. The background window can be changed with the "Background" slider, or set back to the default one with "Reset to default".

```python
tool.background_selection()
```

## Onsets

Determines the onsets of every grouped channel and viewing of each spacecraft, and selects the ones used for the VDA. The number of grouped channels with an onset is printed for each spacecraft. With the "Interactive" onset selection, choose the spacecraft, then the channel with the dropdown or the "Previous" / "Next" buttons, and the viewing whose onset is used (or "None" to leave the channel out).

```python
tool.run_onsets()
```

## VDA

Fits the VDA line of each spacecraft and shows:

- **the results** of each spacecraft:
  - the release time at the Sun, where the fit meets inverse beta 0, with its error (1σ)
  - the times its light reaches 1 AU (for comparison with observations at Earth) and the spacecraft (with the light travel time from the Sun)
  - the apparent path length (APL, in AU)
  - a warning if the fit is not physical: check the background windows and the onsets
- **the combined plot**, saved as a .png file if selected in the Views/Plots tab:
  - the onsets of each spacecraft in its color, of the proton channels as circles and of the electron channels as stars
  - the fit line and 2σ band of each spacecraft from inverse beta 0, where an open circle marks its release time at the Sun
  - the reference times
  - a table under the plot with the release time at the Sun, the time at 1 AU, their error and the APL of each spacecraft
- **the VDA plot of each spacecraft**, if selected in the Views/Plots tab

The results are in the `tool.results` table, one row per spacecraft.

```python
tool.run_vda()
```
