# SPEARHEAD VDA tool: custom points

The tool performs the Velocity Dispersion Analysis (VDA) of onset points given by the user and outputs the release time and apparent path length of each group of points, with their VDA fits in one plot.

A point is an onset time, the energy of its channel (a single energy or a range, in MeV) and its particle (protons or electrons). The points are in groups, e.g. one per spacecraft, each with the distance of its observer from the Sun. The distance gives only the light travel time from the Sun to the observer, for the release time at the observer: the release time at the Sun, the time at 1 AU and the APL do not depend on it. The example points are those of the 2024-11-21 event seen by SOHO, STEREO-A and Solar Orbiter (`examples/custom_points_example.csv`).

Run the cells in order: set the points in the form, then run the VDA. If a point is changed, rerun the VDA. See the [user guide](https://github.com/spearhead-he/VDA/blob/main/docs/user_guide.md#custom-points) for details.

```python
try:
    from spearhead.vda.notebook import VDA_custom_points_notebook
except ModuleNotFoundError:
    # running from the repository without installing the package
    import sys
    sys.path.insert(0, "src")
    from spearhead.vda.notebook import VDA_custom_points_notebook

tool = VDA_custom_points_notebook()
# the example points, to be changed in the form or replaced with tool.load_points("path.csv")
tool.load_points("examples/custom_points_example.csv")
```

## Parameters

- **Points**: a .csv points file, or the groups of points if no file is given. The file has one row per point, with the columns `Group` (optional), `Distance` (AU, the same for all the points of a group), `Particle` (protons or electrons), `Energy` (MeV, a single energy, e.g. `13.5`, or a range, e.g. `13-16`) and `Onset Time` (see `examples/custom_points_example.csv`). In the form, each group is a tab with its label, distance from the Sun and points ("Add Point" / "Remove"), and groups are added with "Add Group". The energy of a range is its geometric mean.
- **Views/Plots**:
  - font sizes, and saving of the VDA plot as a .png file
  - the title of the plot
  - reference times (e.g. SXR peak, type II and type III radio bursts): a label, the observed datetime and the observer (Earth or a spacecraft, or its distance from the Sun). They are shown at inverse beta 0 at the Sun, i.e. minus the light travel time from the Sun to the observer

```python
tool.parameters_form()
```

## VDA

Fits the VDA line of each group and shows:

- **the results** of each group:
  - the release time at the Sun, where the fit meets inverse beta 0, with its error (1σ)
  - the times its light reaches 1 AU (for comparison with observations at Earth) and the observer (with the light travel time from the Sun)
  - the apparent path length (APL, in AU)
  - a warning if the fit is not physical: check the background windows and the onsets
- **the VDA plot**, saved as a .png file if selected in the Views/Plots tab:
  - the points of each group in its color, protons as circles and electrons as stars
  - the fit line and 2σ band of each group from inverse beta 0, where an open circle marks its release time at the Sun
  - the reference times
  - a table under the plot with the release time at the Sun, the time at 1 AU, their error and the APL of each group

The results are in the `tool.results` table, one row per group.

```python
tool.run_vda()
```
