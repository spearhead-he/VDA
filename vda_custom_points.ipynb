# SPEARHEAD VDA tool: custom points

The tool performs the Velocity Dispersion Analysis (VDA) of onset points given by the user and outputs the release time and apparent path length of each group of points, with their VDA fits in one plot.

A point is an onset time, the energy of its channel (a single energy or a range, in MeV) and its particle (protons or electrons). The points are in groups, e.g. one per spacecraft, each with the distance of its observer from the Sun, which gives the light travel time from the Sun (extra time) and the APL. The example points are those of the 2024-11-21 event seen by SOHO, STEREO-A and Solar Orbiter (`examples/custom_points_example.csv`).

Run the cells in order: set the points in the form, then run the VDA. If a point is changed, rerun the VDA. See the [user guide](docs/user_guide.md#custom-points) for details.

```python
try:
    from spearhead.vda.notebook import VDA_custom_points_notebook
except ModuleNotFoundError:
    # running from the repository without installing the package
    import sys
    sys.path.insert(0, "src")
    from spearhead.vda.notebook import VDA_custom_points_notebook

tool = VDA_custom_points_notebook()
# the example points, to be changed in the form or replaced with tool.load_points("path.csv")
tool.load_points("examples/custom_points_example.csv")
```

## Parameters

- **Points**: a .csv points file, or the groups of points if no file is given. The file has one row per point, with the columns `Group` (optional), `Distance` (AU, the same for all the points of a group), `Particle` (protons or electrons), `Energy` (MeV, a single energy, e.g. `13.5`, or a range, e.g. `13-16`) and `Onset Time` (see `examples/custom_points_example.csv`). In the form, each group is a tab with its label, distance from the Sun and points ("Add Point" / "Remove"), and groups are added with "Add Group". The energy of a range is its geometric mean.
- **Views/Plots**: font sizes, saving of the VDA plot, its title, and reference times (e.g. SXR peak, type II and type III radio bursts, with a label, the observed datetime and the observer, Earth or a spacecraft, or its distance from the Sun) shown at inverse beta 0 at the Sun, i.e. minus the light travel time from the Sun to the observer.

```python
tool.parameters_form()
```

## VDA

For each group, the release time (e.g. `15:26:11 (+0:06:40)`: it includes the light travel time from the Sun to the observer, shown in parentheses), the extra time (light travel time from the Sun to the observer, at the distance of the group) and the apparent path length (APL, in AU) of the VDA fit are printed, with a warning for a fit that is not physical, followed by the VDA plot: the points and the VDA fit of each group in its color (protons as circles, electrons as stars), extended to inverse beta 0, where an open circle marks its release time at the Sun, with the reference times. The table under the plot gives the release time at the Sun, the time its light reaches 1 AU (for comparison with observations at Earth), their error and the APL of each group. It is saved as a .png file if selected in the Views/Plots tab. The results are in the `tool.results` table.

```python
tool.run_vda()
```
